# Steps in data cleaning :
1. Inspection(already done in data understanding file).
2. Remove duplicates.
3. Handle missing values
4. Correct data types
5. Standardize text data.
6. Fix inconsistent categories like 'delhi' and 'Delhi'.
7. Handle invalid values.
8. Detect and handle outliers.
9. Remove unnecessary columns.

```python
import numpy as np
import pandas as pd
from google.colab import files
```

# Cleaning the Customer Flight Activity dataset

```python
activity = pd.read_csv('/content/Customer Flight Activity.csv')
activity.head(3)
```

# 1. Missing Values

```python
activity.info()
# no missing values.
```

# 2. Remove Duplicates

```python
activity = activity.drop_duplicates()
activity.duplicated().sum()
```

# 3. Correct datatypes

```python
activity.info()
# no incorrect data types
```

# 4. Handle invalid values

```python
print((activity['Distance'] < 0).sum())
print((activity['Total Flights'] < 0).sum())
# No invalid values
```

# 5. Remove Unnecessary Columns
  'Dollar Cost Points Redeemed' column is unnecessary
  and therefore we will remove it.

```python
activity = activity.drop(columns = "Dollar Cost Points Redeemed")
activity.head(3)
```

# 6. Outlier detection for numerical columns

```python
numeric_cols = activity[['Total Flights', 'Distance', 'Points Accumulated', 'Points Redeemed']]
outlier_summary = []
for col in numeric_cols :
  Q1 = activity[col].quantile(0.25)
  Q3 = activity[col].quantile(0.75)
  IQR = Q3 - Q1
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR
  outliers = ((activity[col] < lower) | (activity[col] > upper)).sum()
  outlier_summary.append({
      'Column' : col,
      'lower_bound' : lower,
      'upper_bound' : upper,
      'outliers' : outliers,
      'percentage' : round(outliers * 100 / len(activity), 2)
  })
outlier_summary = pd.DataFrame(outlier_summary)
outlier_summary
```

# 7. Handling outlier
Outliers were identified using the Interquartile Range (IQR) method. Although several numerical variables contained observations outside the IQR bounds, these values correspond to legitimate high-value customers (e.g., frequent flyers with high travel distances and loyalty point balances). Since these represent meaningful business behavior rather than data errors, no outliers were removed.

# 8. Saving the cleaned File

```python
activity.to_csv('customer_flight_activity_cleaned.csv', index = False)
files.download('customer_flight_activity_cleaned.csv')
```

# Cleaning the Customer Loyalty History dataset

```python
loyalty = pd.read_csv('/content/Customer Loyalty History.csv')
loyalty.head(3)
```

# 1. Remove Duplicates

```python
loyalty = loyalty.drop_duplicates()
loyalty.duplicated().sum()
```

# 2. Handling missing values

```python
loyalty.info()
```

```python
loyalty.groupby('Gender')['Salary'].apply(lambda x: x.isna().sum())
```

```python
loyalty.groupby('Education')['Salary'].apply(lambda x: x.isna().sum())
```

```python
loyalty.groupby('Education')['Salary'].apply(lambda x: x.isna().mean() * 100)
```

```python
loyalty['Salary'] = loyalty['Salary'].fillna(loyalty['Salary'].median())
loyalty['Cancellation Year'] = loyalty['Cancellation Year'].fillna(0)
loyalty['Cancellation Month'] = loyalty['Cancellation Month'].fillna(0)
loyalty.info()
```

There 4238 missing values in the Salary and interestingly all of them belonged to the same education grouop 'College'. This looks like a data collection issue. The missing values were replaced with the median of the whole salary column.
Naturally, there were missing values in the Cancellation Year and Cancellation Month columns which were replaced by 0.

# 3. Standardize Text Data

```python
text_cols = loyalty.select_dtypes(include = 'object').columns
loyalty[text_cols] = loyalty[text_cols].apply(lambda x : x.str.strip())
loyalty[text_cols] = loyalty[text_cols].apply(lambda x : x.str.lower())
loyalty[text_cols]
```

# 4. Handling invalid values

```python
print((loyalty['Salary'] < 0).sum())
loyalty[loyalty['Salary'] < 0]
```

It's most likely a data entry error so we will replace them with the absolute value

```python
loyalty['Salary'] = abs(loyalty['Salary'])
print((loyalty['Salary'] < 0).sum())
```

# 5. Detecting Outliers

```python
numeric_cols = ['Salary', 'CLV']
loyalty[numeric_cols]
outlier_summary = []
for col in numeric_cols :
  Q1 = loyalty[col].quantile(0.25)
  Q3 = loyalty[col].quantile(0.75)
  IQR = Q3 - Q1
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR
  outliers = ((loyalty[col] < lower) | (loyalty[col] > upper)).sum()
  outlier_summary.append({
      'Columns' : col,
      'lower_bound' : lower,
      'Upper_bound' : upper,
      'outliers' : outliers,
      'percentage' : round(outliers * 100 / len(loyalty), 2)
  })
outlier_summary = pd.DataFrame(outlier_summary)
outlier_summary
```

As before, The outliers are genuine data variations and therefore are kept intact.

# 6. Saving the File

```python
loyalty.to_csv('customer_loyalty_history.csv', index = False)
files.download('customer_loyalty_history.csv')
```
